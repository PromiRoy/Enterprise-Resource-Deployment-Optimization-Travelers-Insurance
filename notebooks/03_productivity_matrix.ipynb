```python
import pandas as pd
import numpy as np

claims_per_day = {
    5: {5: 0.25, 4: 0.125, 3: np.nan, 2: np.nan, 1: np.nan},
    4: {5: 1.00, 4: 0.75,  3: 0.125, 2: np.nan, 1: np.nan},
    3: {5: 2.00, 4: 2.00,  3: 2.00,  2: 1.50,  1: np.nan},
    2: {5: 2.50, 4: 2.50,  3: 2.50,  2: 2.00,  1: 0.50},
    1: {5: 3.00, 4: 3.00,  3: 3.00,  2: 3.00,  1: 3.00},
}

rows = []
for severity, skills in claims_per_day.items():
    for skill_level, cpd in skills.items():
        if pd.isna(cpd):
            continue
        rows.append({
            "CAT Severity Code": int(severity),
            "Skill Level": int(skill_level),
            "service_days": round(1.0 / float(cpd), 3),
        })

ds_service_days = (
    pd.DataFrame(rows)
      .sort_values(["CAT Severity Code", "Skill Level"], ascending=[False, False])
      .reset_index(drop=True)
)

ds_service_days
```

```python
ds_service_days.to_excel(
    "../data/processed/ds_productivityMatrix.xlsx",
    index=False
)
```
